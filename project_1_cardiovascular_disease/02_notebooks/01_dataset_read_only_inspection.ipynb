# Cardiovascular Dataset Read-Only Inspection

This notebook performs the first inspection of the raw cardiovascular disease dataset.

It is intentionally read-only:

* it does not clean the dataset
* it does not save a processed dataset
* it does not train a model
* it does not create app files

The purpose is to confirm whether the raw data is suitable for the planned data science project and to identify the cleaning rules that should be reviewed before preparation begins.

## Inspection Objectives

This notebook checks:

* raw file path and source
* delimiter handling
* row and column count
* column names and data types
* missing values
* duplicate records
* target distribution
* basic category values
* age, BMI, height, weight, and blood-pressure quality flags
* initial suitability and cleaning recommendations

```python
from pathlib import Path

import numpy as np
import pandas as pd

pd.set_option("display.max_columns", 50)
pd.set_option("display.width", 120)


def find_project_dir() -> Path:
    """Find the cardiovascular project folder from common VS Code working directories."""
    cwd = Path.cwd().resolve()
    if cwd.name == "02_notebooks":
        return cwd.parent
    if cwd.name == "project_1_cardiovascular_disease":
        return cwd
    for candidate in [cwd, *cwd.parents]:
        if candidate.name == "project_1_cardiovascular_disease":
            return candidate
    child_project = cwd / "project_1_cardiovascular_disease"
    if child_project.exists():
        return child_project
    raise FileNotFoundError("Could not find project_1_cardiovascular_disease from the current working directory.")


def add_inspection_fields(dataframe: pd.DataFrame) -> pd.DataFrame:
    """Add derived fields used only for read-only inspection checks."""
    output = dataframe.copy()
    output["age_years"] = output["age"] / 365.25
    output["bmi"] = output["weight"] / ((output["height"] / 100) ** 2)
    return output


def ensure_raw_data():
    """Return project path, raw file path, and the raw dataset."""
    project_dir = find_project_dir()
    raw_path = project_dir / "01_data" / "raw" / "cardio_data.csv"
    if not raw_path.exists():
        raise FileNotFoundError(f"Raw dataset not found: {raw_path}")
    raw_data = pd.read_csv(raw_path, sep=";")
    return project_dir, raw_path, raw_data


PROJECT_DIR, RAW_PATH, raw_df = ensure_raw_data()

file_check = pd.DataFrame([
    {
        "raw_path": str(RAW_PATH.relative_to(PROJECT_DIR)),
        "exists": RAW_PATH.exists(),
        "size_mb": round(RAW_PATH.stat().st_size / (1024 * 1024), 2),
    }
])

file_check
```

```python
# This guard lets the cell run cleanly even if previous cells were skipped.
if "ensure_raw_data" not in globals():
    from pathlib import Path

    import numpy as np
    import pandas as pd

    pd.set_option("display.max_columns", 50)
    pd.set_option("display.width", 120)


    def find_project_dir() -> Path:
        """Find the cardiovascular project folder from common VS Code working directories."""
        cwd = Path.cwd().resolve()
        if cwd.name == "02_notebooks":
            return cwd.parent
        if cwd.name == "project_1_cardiovascular_disease":
            return cwd
        for candidate in [cwd, *cwd.parents]:
            if candidate.name == "project_1_cardiovascular_disease":
                return candidate
        child_project = cwd / "project_1_cardiovascular_disease"
        if child_project.exists():
            return child_project
        raise FileNotFoundError("Could not find project_1_cardiovascular_disease from the current working directory.")


    def add_inspection_fields(dataframe: pd.DataFrame) -> pd.DataFrame:
        """Add derived fields used only for read-only inspection checks."""
        output = dataframe.copy()
        output["age_years"] = output["age"] / 365.25
        output["bmi"] = output["weight"] / ((output["height"] / 100) ** 2)
        return output


    def ensure_raw_data():
        """Return project path, raw file path, and the raw dataset."""
        project_dir = find_project_dir()
        raw_path = project_dir / "01_data" / "raw" / "cardio_data.csv"
        if not raw_path.exists():
            raise FileNotFoundError(f"Raw dataset not found: {raw_path}")
        raw_data = pd.read_csv(raw_path, sep=";")
        return project_dir, raw_path, raw_data


    PROJECT_DIR, RAW_PATH, raw_df = ensure_raw_data()

PROJECT_DIR, RAW_PATH, raw_df = ensure_raw_data()

inspection_df = add_inspection_fields(raw_df)

eda_signal_df = inspection_df.copy()
eda_signal_df["age_band"] = pd.cut(eda_signal_df["age_years"], bins=[0, 40, 45, 50, 55, 60, 70], right=False)
eda_signal_df["bmi_band"] = pd.cut(eda_signal_df["bmi"], bins=[0, 18.5, 25, 30, 35, 40, 100], right=False)
eda_signal_df["systolic_band"] = pd.cut(eda_signal_df["ap_hi"], bins=[0, 120, 130, 140, 160, 1000], right=False)

signal_tables = []
for column in ["age_band", "bmi_band", "systolic_band", "cholesterol", "gluc", "active"]:
    summary = (
        eda_signal_df
        .groupby(column, observed=False)["cardio"]
        .agg(rows="count", cardio_rate="mean")
        .reset_index()
    )
    summary.insert(0, "feature", column)
    summary = summary.rename(columns={column: "category"})
    signal_tables.append(summary)

eda_signal_summary = pd.concat(signal_tables, ignore_index=True)
eda_signal_summary["cardio_rate"] = eda_signal_summary["cardio_rate"].round(3)
eda_signal_summary
```
